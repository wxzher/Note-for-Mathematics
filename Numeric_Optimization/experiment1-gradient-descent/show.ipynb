```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

# 定义 Modified Rosenbrock 函数
def modified_rosenbrock(x1, x2):
    part1 = (1 - x1)**2
    part2 = 100 * (x2 - x1**2)**2
    part3 = 5 * np.sin(2 * np.pi * x1) * np.sin(2 * np.pi * x2)
    return part1 + part2 + part3

# 设置绘图网格
x1 = np.linspace(-2, 2, 400)
x2 = np.linspace(-1, 3, 400)
X1, X2 = np.meshgrid(x1, x2)
Z = modified_rosenbrock(X1, X2)

# 图1：3D 曲面图
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(X1, X2, Z, cmap=cm.viridis, edgecolor='none', alpha=0.9)
ax.view_init(elev=30, azim=-60)
ax.set_title('Modified Rosenbrock Function 3D Surface', fontsize=14)
ax.set_xlabel('$x_1$')
ax.set_ylabel('$x_2$')
ax.set_zlabel('$f(x)$')
fig.colorbar(surf, shrink=0.5, aspect=5)
plt.savefig('modified_rosenbrock_3d.png', dpi=300, bbox_inches='tight')
plt.show()

# 图2：等高线图 (展示多极值特性)
plt.figure(figsize=(8, 6))
# 使用对数刻度使得山谷结构更清晰
levels = np.logspace(-1, 3.5, 30)
cp = plt.contour(X1, X2, Z, levels=levels, cmap='jet')
plt.colorbar(cp, label='Function Value (Log Scale generally)')
plt.title('Modified Rosenbrock Contour (Log Levels)', fontsize=14)
plt.xlabel('$x_1$')
plt.ylabel('$x_2$')
# 标记全局极小值大概位置 (1, 1) 注意：加上正弦项后略有偏移，但大体在此
plt.plot(1, 1, 'r*', markersize=15, label='Global Min (Approx)')
plt.legend()
plt.savefig('modified_rosenbrock_contour.png', dpi=300, bbox_inches='tight')
plt.show()
```
